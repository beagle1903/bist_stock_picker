```python
import yfinance as yf
import pandas as pd
import numpy as np
from tqdm import tqdm
```

```python
bist100 = [
    "AEFES.IS","AGHOL.IS","AKBNK.IS","AKSA.IS","AKSEN.IS","ALARK.IS","ASELS.IS",
    "BIMAS.IS","BRISA.IS","CCOLA.IS","CIMSA.IS","DOAS.IS","DOHOL.IS",
    "EKGYO.IS","ENKAI.IS","EREGL.IS","FROTO.IS","GARAN.IS","GUBRF.IS",
    "HEKTS.IS","ISCTR.IS","KCHOL.IS","KRDMD.IS",
    "MGROS.IS","ODAS.IS","OTKAR.IS","PETKM.IS","PGSUS.IS","SAHOL.IS",
    "SASA.IS","SISE.IS","TAVHL.IS","TCELL.IS","THYAO.IS","TKFEN.IS",
    "TOASO.IS","TSKB.IS","TUPRS.IS","ULKER.IS","VESTL.IS","YKBNK.IS"
]
```

```python
data = yf.download(
    bist100,
    period="3mo",
    interval="1d",
    auto_adjust=True,
    group_by="ticker",
    threads=True
)
```

```python
data
```

```python
performance = {}

for ticker in bist100:
    try:
        df = data[ticker].dropna()
        if len(df) > 0:
            start_price = df["Close"].iloc[0]
            end_price = df["Close"].iloc[-1]
            perf = (end_price / start_price - 1) * 100
            performance[ticker] = perf
    except:
        pass

perf_df = pd.DataFrame.from_dict(performance, orient="index", columns=["3M_%"])
perf_df = perf_df.sort_values("3M_%", ascending=False)

perf_df.head(20)
```

```python
perf_df.to_csv("bist100_3m_performance.csv")
```

```python
bist100 = [
"AEFES.IS","AGHOL.IS","AHGAZ.IS","AKBNK.IS","AKCNS.IS","AKFGY.IS","AKSA.IS","AKSEN.IS","ALARK.IS","ALBRK.IS",
"ALFAS.IS","ARCLK.IS","ASELS.IS","ASTOR.IS","BERA.IS","BIMAS.IS","BRSAN.IS","BRYAT.IS","BUCIM.IS","CCOLA.IS",
"CIMSA.IS","DOAS.IS","DOHOL.IS","ECILC.IS","EGEEN.IS","EKGYO.IS","ENJSA.IS","ENKAI.IS","EREGL.IS","EUPWR.IS",
"FROTO.IS","GARAN.IS","GESAN.IS","GUBRF.IS","GWIND.IS","HALKB.IS","HEKTS.IS","ISCTR.IS","ISDMR.IS","ISGYO.IS",
"KCAER.IS","KCHOL.IS","KLSER.IS","KONTR.IS","KOZAA.IS","KOZAL.IS","KRDMD.IS","MAVI.IS","MGROS.IS","ODAS.IS",
"OTKAR.IS","OYAKC.IS","PETKM.IS","PGSUS.IS","QUAGR.IS","SAHOL.IS","SASA.IS","SDTTR.IS","SISE.IS","SKBNK.IS",
"SMRTG.IS","SOKM.IS","TABGD.IS","TAVHL.IS","TCELL.IS","THYAO.IS","TKFEN.IS","TKNSA.IS","TMSN.IS","TOASO.IS",
"TSKB.IS","TTRAK.IS","TUPRS.IS","TURSG.IS","ULKER.IS","VAKBN.IS","VESBE.IS","VESTL.IS","YATAS.IS","YKBNK.IS",
"YYLGD.IS","ZOREN.IS"
]
```

```python
def calc_performance(df, months):
    if len(df) < 5:
        return np.nan
    days = int(months * 21)  # approx trading days
    if len(df) <= days:
        start_price = df["Close"].iloc[0]
    else:
        start_price = df["Close"].iloc[-days]
    end_price = df["Close"].iloc[-1]
    return (end_price / start_price - 1) * 100
```

```python
results = []

for ticker in tqdm(bist100):
    try:
        stock = yf.Ticker(ticker)
        hist = stock.history(period="7mo", auto_adjust=True).dropna()

        if len(hist) == 0:
            continue

        perf_6m = calc_performance(hist, 6)
        perf_3m = calc_performance(hist, 3)
        perf_1m = calc_performance(hist, 1)

        info = stock.info
        sector = info.get("sector", "Unknown")
        industry = info.get("industry", "Unknown")

        results.append({
            "Ticker": ticker.replace(".IS",""),
            "6M_%": round(perf_6m, 2),
            "3M_%": round(perf_3m, 2),
            "1M_%": round(perf_1m, 2),
            "Sector": sector,
            "Industry": industry
        })

    except Exception as e:
        print("Error:", ticker, e)
```

```python
df = pd.DataFrame(results)
df = df.sort_values("6M_%", ascending=False)
df.head()
```

```python
df.to_csv("bist100_performance.csv", index=False)
print("CSV saved: bist100_performance.csv")
```
